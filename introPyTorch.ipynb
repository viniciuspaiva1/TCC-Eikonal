```python
class LineNetwork(nn.Module):
    def __init__(self, *args, **kwargs):
        #Inicialização da clase mãe 
        super().__init__(*args, **kwargs)

        self.layers = nn.Sequential(
            nn.Linear(1,1)
        )
    
    def foward(self, x):
        #aplica os dados X ao neuronio
        return self.layers(x)
```

## Preparando os dados


```python
from torch.utils.data import Dataset, DataLoader
import torch.distributions.uniform as urand
```

```python
class AlgebraicDataSet(Dataset):
    def __init__(self, function, interval, nsamples):
        X = urand.Uniform(interval[0], interval[1]).sample([nsamples])
        self.data = [(x, function(x)) for x in X]

    def __len__(self):
        return len(self.data)

    def __getitem__(self, index):
        return self.data[index]
```

```python
f_line = lambda x: 2*x+3
interval = [-10,10]
train_nsamples = 1000
test_nsamples = 100
```

```python
train_dataset = AlgebraicDataSet(f_line,interval, train_nsamples)
test_dataset = AlgebraicDataSet(f_line,interval, test_nsamples)

train_dataloader = DataLoader(train_dataset, batch_size=train_nsamples, shuffle=True)
test_dataloader = DataLoader(train_dataset, batch_size=test_nsamples, shuffle=True)
```
