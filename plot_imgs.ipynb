```python
import numpy as np
import matplotlib.pyplot as plt

# Conjunto A: todos os pares (x, z) de 0 a 5
x_a = np.arange(0, 6)
z_a = np.arange(0, 6)
X_a, Z_a = np.meshgrid(x_a, z_a)
X_a = X_a.flatten()
Z_a = Z_a.flatten()

# Conjunto B: pares com step 2
x_b = np.arange(0, 31, 2)
z_b = np.arange(0, 31, 2)
X_b, Z_b = np.meshgrid(x_b, z_b)
X_b = X_b.flatten()
Z_b = Z_b.flatten()

# Plot
fig, axes = plt.subplots(1, 2, figsize=(10, 5))

# Conjunto A
axes[0].scatter(X_a, Z_a, c='blue')
axes[0].set_title('Conjunto do tipo I')
axes[0].set_xticks(np.arange(0, 31, 5))
axes[0].set_yticks(np.arange(0, 31, 5))
axes[0].grid(False)
axes[0].set_xlabel('X')
axes[0].set_ylabel('Z')
axes[0].set_aspect('equal')
axes[0].invert_yaxis()  # <<< Aqui inverte o eixo Z (vertical)

# Conjunto B
axes[1].scatter(X_b, Z_b, c='red')
axes[1].set_title('Conjunto do tipo II(step=2)')
axes[1].set_xticks(np.arange(0, 31, 5))
axes[1].set_yticks(np.arange(0, 31, 5))
axes[1].grid(False)
axes[1].set_xlabel('X')
axes[1].set_ylabel('Z')
axes[1].set_aspect('equal')
axes[1].invert_yaxis()  # <<< Aqui inverte o eixo Z (vertical)

plt.tight_layout()
plt.show()
```
